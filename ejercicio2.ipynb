# Construir el siguiente DataFrame

```python
import pandas as pd

datos = {
" Producto ":
[" Mouse "," Teclado ",
" Monitor "," Laptop "] ,
" Precio ":
[250 ,450 ,3200 ,15000] ,
" Existencia ":
[20 ,15 ,8 ,5]
}

df = pd.DataFrame(datos)
print(df)
```

# Columna precio

```python
precio=df[" Precio "]
print(precio)
```

# Mostrar productos mayores a 500

```python
# Mostrar productos mayores a 500
df_mayor_500 = df[df[" Precio "] > 500]
print(df_mayor_500)
```

# Crear columna con iva

```python
# 3 Crear la columna IVA.
df["IVA"] = df[" Precio "] * 0.19
print(df)
```

# Crear la columna precio final

```python
# 4 Crear la columna Precio Final.
df["Precio Final"] = df[" Precio "] + df["IVA"]
print(df)
```

# Mostrar los que tengan existencia menor a 10

```python
# Mostrar ´unicamente los productos con existencia menor a 10.
df_existencia_menor_10 = df[df[" Existencia "] < 10]
print(df_existencia_menor_10)
```
