# Ejercicio 1: Registrador de Estudiantes
Escribe una función llamada `crear_estudiante(nombre, edad, carrera)` que reciba estos tres datos como parámetros y devuelva un diccionario con la siguiente estructura:


*Ejemplo de uso:*
`alumno = crear_estudiante("Daniel", 21, "Computación")`
`print(alumno)`   

*Salida esperada:*
`{'nombre': 'Daniel', 'edad': 21, 'carrera': 'Computación', 'activo': True}`

```python
def crear_estudiante(nombre, edad, carrera):
    estudiante = {
        "nombre": nombre,
        "edad": edad,
        "carrera": carrera
    }
    return estudiante

# Ejemplo de uso
alumno1 = crear_estudiante("Juanito", 20, "Ingeniería Informática")
print(alumno1)
```

# Ejercicio 2: Consultar Precios (Manejo de errores con .get()) 
Crea una función llamada `obtener_precio(inventario, producto)` que reciba un diccionario de precios y el nombre de un producto a buscar.
Si el producto existe en el inventario, debe devolver: "El precio de `[producto]` es `$[precio]`".
Si el producto no existe, debe devolver: "El producto `[producto]` no está disponible".

*Ejemplo de entrada:*
`precios = {"manzana": 15.5, "pila": 25.0, "cuaderno": 45.0}`

*Pruebas:*
`print(obtener_precio(precios, "pila"))`     # -> El precio de pila es $25.0
`print(obtener_precio(precios, "lapicero"))` # -> El producto lapicero no está disponible


```python
def obtener_precio(inventario, producto):
    precio = inventario.get(producto)
    if precio is not None:
        return f"El precio de {producto} es ${precio}"
    else:
        return f"El producto {producto} no está disponible en el inventario."

precios = {"manzana": 15.5, "pila": 25.0, "cuaderno": 45.0}
print(obtener_precio(precios, "pila"))
print(obtener_precio(precios, "lapicero"))
```

# Ejercicio 3: Contadores de Frecuencia
Escribe una función llamada `contar_caracteres(texto)` que reciba una cadena de texto y devuelva un diccionario donde las claves sean las letras y los valores la cantidad de veces que aparece cada una.

*Ejemplo de uso:*
`resultado = contar_caracteres("programacion")`
`print(resultado)`

*Salida esperada:*
`{'p': 1, 'r': 2, 'o': 2, 'g': 1, 'a': 2, 'm': 1, 'c': 1, 'i': 1, 'n': 1}`

```python
def contar_caracteres(cadena):
    conteo = {}
    for caracter in cadena:
        if caracter in conteo:
            conteo[caracter] += 1
        else:
            conteo[caracter] = 1
    return conteo

resultado = contar_caracteres("programacion")
print(resultado)
```

# Ejercicio 4: Filtrado de Calificaciones
Crea una función llamada `filtrar_aprobados(calificaciones, nota_minima)` que reciba un diccionario con los nombres de los alumnos y sus notas, junto con la nota mínima para aprobar (por defecto 6.0). Debe devolver un nuevo diccionario únicamente con los alumnos aprobados.

*Ejemplo de uso:*
`grupo = {"Ana": 8.5, "Carlos": 5.0, "Sofia": 9.2, "Luis": 5.8, "Diana": 7.0}`

`aprobados = filtrar_aprobados(grupo, nota_minima=6.0)`
`print(aprobados)`

*Salida esperada:*
`{'Ana': 8.5, 'Sofia': 9.2, 'Diana': 7.0}`
`print(obtener_precio(precios, "lapicero"))` # -> El producto lapicero no está disponible

```python
def filtrar_aprobados(calificaciones, nota_minima):
    aprobados = {}
    for estudiante, nota in calificaciones.items():
        if nota >= nota_minima:
            aprobados[estudiante] = nota
    return aprobados

grupo_calificaciones = {"Ana": 8.5, "Luis": 7.0, "Marta": 9.0, "Carlos": 6.0, "Sofía": 9.5}
aprobados = filtrar_aprobados(grupo_calificaciones, 7.5)
print(aprobados)
```
