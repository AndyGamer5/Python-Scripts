# Ejercicio 1: Registrador de Estudiantes
Escribe una función llamada `crear_estudiante(nombre, edad, carrera)` que reciba estos tres datos como parámetros y devuelva un diccionario con la siguiente estructura:


*Ejemplo de uso:*
`alumno = crear_estudiante("Daniel", 21, "Computación")`
`print(alumno)`   

*Salida esperada:*
`{'nombre': 'Daniel', 'edad': 21, 'carrera': 'Computación', 'activo': True}`

```python
def crear_estudiante(nombre, edad, carrera):
    estudiante = {
        "nombre": nombre,
        "edad": edad,
        "carrera": carrera
    }
    return estudiante

# Ejemplo de uso
alumno1 = crear_estudiante("Juanito", 20, "Ingeniería Informática")
print(alumno1)
```
