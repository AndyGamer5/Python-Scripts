## Primera Clase1:

1. Enteros(`int`), Flotantes(`float`)
2. Cadenas(`str`), Listas(`list`), conjuntos(`set`).

Que son estos objetos?

Como **conjunto**, los enteros son de la siguente forma:
$$\mathbb{Z}=\{\dots, -1, -2, -0, 1, 2, \cdots \}$$

Un **string** es una concatenacion de caracteres: `"Hola a todos!"`

Un ejemplo de una **lista** es la forma:
$$L = [ l_{0}, l_{1}, l_{2}, \cdots, l_{n} ]$$
donde cada $l_{i}$ es, a su vez, un objeto para cada $i=0, \cdots, n$
