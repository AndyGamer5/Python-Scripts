```python
import random

def simular_lanzamientos(cantidad):
    eventos = [1, 2, 3, 4, 5, 6]  # Representa los posibles resultados de un dado de seis caras

    resultados = [random.choice(eventos) for _ in range(cantidad)]

    print(f"--- Resultados de  {cantidad} lanzamientos de un dado ---")
    for i, resultado in enumerate(resultados, 1):
        print(f"Lanzamiento {i}: {resultado}")

    # Contar las veces que aparece cada número
    conteo = {i: resultados.count(i) for i in set(resultados)}
    for numero, cantidad in conteo.items():
        print(f"El número {numero} salió {cantidad} veces.")

simular_lanzamientos(5)
```
