 **Ejercicio 1: Procesamiento de Diccionarios Anidados (25 pts)**<br>

Dada la siguiente estructura de datos que representa una base de datos escolar:<br>
`escuela = {
2 " A101 ": {" nombre ": "Ana ", " calificaciones ": {" Matematicas ": 85 , "
Fisica ": 90 , " Programacion ": 95}} ,
3 " A102 ": {" nombre ": " Luis ", " calificaciones ": {" Matematicas ": 78 , "
Fisica ": 82 , " Programacion ": 80}} ,
4 " A103 ": {" nombre ": " Pedro ", " calificaciones ": {" Matematicas ": 92 , "
Fisica ": 88 , " Programacion ": 98}}
5 }`<br>

Desarrolla un script que realice lo siguiente sin utilizar Pandas:

1. Calcule e imprima el promedio individual de cada estudiante.
2. Encuentre y muestre el nombre del estudiante con el mejor promedio general.
3. Calcule el promedio general de toda la escuela para la materia de “Programacion”.

````python
#base de datos de estudiantes y sus calificaciones
escuela = {
    "A101": {"nombre": "Ana",   "calificaciones": {"Matematicas": 85, "Fisica": 90, "Programacion": 95}},
    "A102": {"nombre": "Luis",  "calificaciones": {"Matematicas": 78, "Fisica": 82, "Programacion": 80}},
    "A103": {"nombre": "Pedro", "calificaciones": {"Matematicas": 92, "Fisica": 88, "Programacion": 98}},
}

# 1. Promedio individual de cada estudiante
promedios = {}
for estudiante_id, datos in escuela.items():
    calificaciones = datos["calificaciones"]
    promedio = sum(calificaciones.values()) / len(calificaciones)
    promedios[estudiante_id] = promedio
    print(f"Promedio de {datos['nombre']}: {promedio:.2f}")


# 2. Estudiante con el mejor promedio general
mejor_id = max(promedios, key=promedios.get)
print(f"\nMejor promedio: {escuela[mejor_id]['nombre']} con {promedios[mejor_id]:.2f}")

# 3. Promedio general de la escuela en "Programacion"
suma_programacion = sum(datos["calificaciones"]["Programacion"] for datos in escuela.values())
promedio_programacion = suma_programacion / len(escuela)
print(f"\nPromedio general de la escuela en Programacion: {promedio_programacion:.2f}")
````

 **Ejercicio 2: Frecuencias e Inversion de Diccionarios (25 pts)**
 
Escribe un programa que reciba una cadena de texto larga (por ejemplo, un parrafo de un
libro). El programa debe:

1. Crear un diccionario donde las claves sean cada palabra unica en el texto (ignorando
mayusculas, minusculas y signos de puntuacion) y los valores sean la cantidad de veces
que aparece esa palabra.
2. Invertir el diccionario resultante: crear un nuevo diccionario donde las claves sean las
frecuencias (n´umeros) y los valores sean listas con las palabras que tienen esa frecuencia.
3. Imprimir las 3 palabras m´as frecuentes.

````python
def contar_palabras(texto):
    palabras = texto.split()
    dict_palabras = {}
    for palabra in palabras:
        palabra = palabra.lower().strip(",.!?;:")
        if palabra in dict_palabras:
            dict_palabras[palabra] += 1
        else:
            dict_palabras[palabra] = 1
    return dict_palabras

texto = "Hola, hola! ¿Cómo estás? Estoy bien, gracias. Hola de nuevo."
conteo = contar_palabras(texto)
print("\nConteo de palabras en el texto:")
for palabra, cantidad in conteo.items():
    print(f"{palabra}: {cantidad}")
````

**Ejercicio 3: Limpieza y Agregacion con Pandas (25 pts)**<br>

Apartir de los siguientes datos que contienen valores nulos:<br>
```
import pandas as pd
import numpy as np
datos_ventas = {
    `"Sucursal": ["Norte", "Sur", "Norte", "Centro", "Sur", "Centro", "
    Norte"],
    "Vendedor": ["Carlos", "Maria", "Jorge", "Ana", np.nan, "Luis", "

    "Ventas": [15000, 22000, np.nan, 18000, 25000, np.nan, 17500],
    "Clientes": [15, 20, 12, 18, 25, 14, 16]
    10 df_ventas = pd.DataFrame(datos_ventas)
```
Realiza las siguientes operaciones:
1. Rellena los valores nulos de la columna Ventas con la mediana de las ventas totales.
2. Elimina las filas donde el nombre del Vendedor sea un valor nulo.
3. Utiliza el m´etodo groupby() para agrupar los datos por Sucursal y calcula el total de
Ventas y el promedio de Clientes por sucursal

````python
import pandas as pd
import numpy as np

datos_ventas = {
    "Sucursal": ["Norte", "Sur", "Norte", "Centro", "Sur", "Centro", "Norte"],
    "Vendedor": ["Carlos", "Maria", "Jorge", "Ana", np.nan, "Luis", "Elena"],
    "Ventas": [15000, 22000, np.nan, 18000, 25000, np.nan, 17500],
    "Clientes": [15, 20, 12, 18, 25, 14, 16]
}
df_ventas = pd.DataFrame(datos_ventas)

print("DataFrame original:")
print(df_ventas)

# 1. Rellenar los valores nulos de "Ventas" con la mediana
mediana_ventas = df_ventas["Ventas"].median()
df_ventas["Ventas"] = df_ventas["Ventas"].fillna(mediana_ventas)
print(f"\nMediana de Ventas usada para rellenar: {mediana_ventas}")

# 2. Eliminar las filas donde "Vendedor" sea nulo
df_ventas = df_ventas.dropna(subset=["Vendedor"])

print("\nDataFrame despues de limpieza:")
print(df_ventas)

# 3. Agrupar por Sucursal: total de Ventas y promedio de Clientes
resumen_sucursal = df_ventas.groupby("Sucursal").agg(
    Total_Ventas=("Ventas", "sum"),
    Promedio_Clientes=("Clientes", "mean")
)

print("\nResumen por Sucursal:")
print(resumen_sucursal)
````

 **Ejercicio 4: Fusion de DataFrames y Funciones Vectorizadas (25 pts)**<br>
 
Se tienen dos DataFrames distintos: uno con la informaci´on de los empleados y otro con el
catalogo de departamentos.<br>
``` 
 empleados = pd.DataFrame({ 
    "ID": [1, 2, 3, 4, 5],
    "Nombre": ["Ana", "Luis", "Pedro", "Marta", "Jorge"],
    "ID_Depto": [10, 20, 10, 30, 20],
    "Salario_Base": [25000, 32000, 28000, 45000, 31000]})`

`departamentos = pd.DataFrame({
    "ID_Depto": [10, 20, 30],
    "Departamento": ["Sistemas", "Ventas", "Direccion"],
    "Bono_Porcentaje": [0.15, 0.25, 0.30]})
```
1. Fusiona (merge) ambos DataFrames utilizando ID
2. Crea una nueva columna llamada Salario
    `Depto como llave.
    Neto, calculada como el Salario
    monto correspondiente al Bono
    Porcentaje.
    Base mas el`
3. Filtra el DataFrame resultante para mostrar ´unicamente a los empleados cuyo Salario
sea superior a $35,000.
4. Exporta este ´ultimo filtro a un archivo llamado empleados
    `Neto
    destacados.csv delimitado
    por punto y coma (sep=’;’).`

````python
import pandas as pd

empleados = pd.DataFrame({
    "ID": [1, 2, 3, 4, 5],
    "Nombre": ["Ana", "Luis", "Pedro", "Marta", "Jorge"],
    "ID_Depto": [10, 20, 10, 30, 20],
    "Salario_Base": [25000, 32000, 28000, 45000, 31000]
})

departamentos = pd.DataFrame({
    "ID_Depto": [10, 20, 30],
    "Departamento": ["Sistemas", "Ventas", "Direccion"],
    "Bono_Porcentaje": [0.15, 0.25, 0.30]
})

# 1. Fusionar ambos DataFrames usando ID_Depto como llave
df_completo = pd.merge(empleados, departamentos, on="ID_Depto", how="left")
print("DataFrame fusionado:")
print(df_completo)

# 2. Crear la columna Salario_Neto (vectorizado, sin apply/loops)
df_completo["Salario_Neto"] = df_completo["Salario_Base"] * (1 + df_completo["Bono_Porcentaje"])

print("\nDataFrame con Salario_Neto:")
print(df_completo)

# 3. Filtrar empleados con Salario_Neto superior a $35,000
empleados_destacados = df_completo[df_completo["Salario_Neto"] > 35000]
print("\nEmpleados destacados (Salario_Neto > $35,000):")
print(empleados_destacados)

# 4. Exportar el filtro a CSV delimitado por punto y coma
empleados_destacados.to_csv("empleados_destacados.csv", sep=";", index=False)
````
